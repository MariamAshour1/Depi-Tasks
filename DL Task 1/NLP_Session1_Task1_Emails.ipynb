## 5 & 6. Create `char_count`, then find the longest and shortest messages

```python
# char_count = number of characters in the message, including spaces,
# punctuation, digits and emojis (each emoji counts as 1 character in Python).
df["char_count"] = df["message"].str.len()

longest  = df.loc[df["char_count"].idxmax()]
shortest = df.loc[df["char_count"].idxmin()]

print("LONGEST  :", longest["message_id"], "|", longest["char_count"], "characters")
print(longest["message"])
print()
print("SHORTEST :", shortest["message_id"], "|", shortest["char_count"], "characters")
print(shortest["message"])
print()
print(df["char_count"].describe().round(1))
```

## What does `char_count` measure?
`char_count` is the total number of characters in the original message. It counts letters, spaces, digits, punctuation and emojis. It is a quick way to see message length, spot very short or very long messages, and later check how much cleaning changed the text.

## Emails: extract into their own column and replace with [EMAIL]
The `emails` column keeps the addresses as a Python list. `clean_message_email` is a new column where each address is replaced by `[EMAIL]`. The original `message` is not changed.

```python
import re

EMAIL_PATTERN = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"

def extract_emails(text):
    return re.findall(EMAIL_PATTERN, text)

def mask_emails(text):
    return re.sub(EMAIL_PATTERN, "[EMAIL]", text)

df["emails"] = df["message"].apply(extract_emails)
df["clean_message_email"] = df["message"].apply(mask_emails)

print("Messages with at least one email:", (df["emails"].str.len() > 0).sum())
df[df["emails"].str.len() > 0][["message_id", "message", "emails", "clean_message_email"]].head(6)
```

**Why replace and not delete?** The email address is private and different for every customer, so it adds noise. But the fact that the customer shared an email is still useful, so we keep a readable placeholder. `@support` is not matched because it has no domain.
