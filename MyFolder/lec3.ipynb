````python
import math 

def area_of_circle(r):
    area = math.pi * (r**2)
    print(f"The area of the circle = {area:.2f}")

area_of_circle(3)

def perimeter_of_circle(r):
    perimeter = 2 * math.pi * r
    print(f"The perimeter of the circle = {perimeter:.2f} ")
    
perimeter_of_circle(5)
    
````

````python
def add_all_nums(*n):
    sum = 0
    for num in n:
        sum += num
    return sum

print(f"The sum of numbers = {add_all_nums(5,3,2,10,15)}")
````

````python
def convert_celsius_2_fahrenheit(c):
    F = (c * (9/5)) + 32 
    return F

print(f"Temprature = {convert_celsius_2_fahrenheit(50)}F")
````

````python
def check_season(month):
    month = month.lower()
    
    if month in ["september", "october", "november"]:
        return "Autumn"
    elif month in ["december", "january", "february"]:
        return "Winter"
    elif month in ["march", "april", "may"]:
        return "Spring"
    elif month in ["june", "july", "august"]:
        return "Summer"
    else:
        return "Invalid month"

season = check_season("JAnuARY")
print(f"The season is {season}")
````

````python
def calculate_slope(x1, y1, x2, y2):
    if x2 - x1 == 0:
        raise ValueError("Vertical line has undefined slope.")
    return (y2 - y1) / (x2 - x1)

print(f"Slope = {calculate_slope(10,6,2,4)}")
````

````python
#$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$

def solve_quadratic_eqn(a,b,c):
    d = b**2 - 4*a*c
    if d < 0:
        return "No real solutions"
    x1 = (-b + math.sqrt(d)) / (2*a)
    x2 = (-b - math.sqrt(d)) / (2*a)
    return (x1, x2) if x1 != x2 else (x1)

    if x1 != x2:
        print(f"Two real solutions found: x1 = {x1:.2f}, x2 = {x2:.2f}")
        return (x1, x2)
    else:
        print(f"One real solution found: x = {x1:.2f}")
        return (x1)
    
solve_quadratic_eqn(1,-5,6)
````

````python
def print_list(listt):
    for item in listt:
        print(f"\n{item}")

listt = ['vegetables', 'apple', 'banana', 'mango']
print_list(listt)
````

````python
def capitalize_list_items(listt):
    return [item.capitalize() for item in listt]

listt = ['vegetables', 'apple', 'banana', 'mango']
print(capitalize_list_items(listt))
````

````python
def add_item(listt, item):
    listt.append(item)
    return listt

listt = ['vegetables', 'apple', 'banana', 'mango']
print(add_item(listt,'yoghurt'))
````

````python
def remove_item(food_staff,item):
    if item in food_staff:
        food_staff.remove(item) 
    return food_staff
food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_staff, 'Mango'))
````

````python
def sum_of_numbers(n):
    total = 0
    for i in range(n+1):
        total += i      
    return total


print(sum_of_numbers(100)) 
````

````python
def sum_of_odds(n):
    total = 0
    for i in range(n+1):
        if i % 2 != 0:
            total += i      
    return total


print(sum_of_odds(100)) 
````

````python
def sum_of_evens(n):
    total = 0
    for i in range(n+1):
        if i % 2 == 0:
            total += i      
    return total


print(sum_of_evens(100)) 
````
